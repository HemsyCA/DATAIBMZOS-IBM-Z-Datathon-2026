```python
import platform, sys
print("Arquitectura:", platform.machine())
print("Python:", sys.version.split()[0])

for lib in ["numpy", "pandas", "sklearn", "torch", "tensorflow"]:
    try:
        m = __import__(lib)
        print(f"✅ {lib}: {m.__version__}")
    except Exception as e:
        print(f"❌ {lib}: {e}")
```
